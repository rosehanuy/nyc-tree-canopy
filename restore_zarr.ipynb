```python
import zarr
import xarray as xr
import pandas as pd
from pathlib import Path
root = Path.cwd()
```

```python
ref_zarr_path = root / 'data' / 'nyc19' / 'nyc19_daily_raw.zarr'

times = pd.to_datetime(xr.open_zarr(ref_zarr_path,consolidated=False)['time'].values)

pd.Series(times).dt.to_period('M').value_counts().sort_index() ## how many time steps each month
```

```python
stop_at = pd.Timestamp('2019-07-01')
zarr_length = int((times<stop_at).sum())

zarr_length
```

```python
store = zarr.open(ref_zarr_path,mode='r+')
store['time'].shape
store['reflectance'].resize((zarr_length,*store['reflectance'].shape[1:]))
store['time'].resize((zarr_length,))
zarr.consolidate_metadata(ref_zarr_path) 
```

```python
## check it worked
times = pd.to_datetime(xr.open_zarr(ref_zarr_path,consolidated=False)['time'].values)
pd.Series(times).dt.to_period('M').value_counts().sort_index()
```
